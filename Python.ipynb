# 7. 파이썬 기본 문법

```python
# 1. 변수와 자료형
# C/Java: int x = 10;  → 자료형(int) 명시 필수
# Python : 자료형 없이 변수명 = 값  → 동적 타이핑(dynamic typing)
# 정적 타이핑(Static Typing) → 변수의 **자료형(type)**을 코드 작성 시점에 미리 지정해야 함.
# 동적 타이핑(Dynamic Typing) → 변수의 자료형이 코드 실행 시점에 값에 따라 자동 결정됨.

x = 10              # 정수형(int)
y = 3.14            # 실수형(float)
name = "Alice"      # 문자열(str)
is_active = True    # 불리언(bool)
print(x, y, name, is_active)
```

```python
# 2. 조건문
# C/Java: 조건문에 () 필수, { }로 블록 표시
# Python : () 불필요, 들여쓰기로 블록 구분
score = 85
if score >= 90:
    print("A학점")
elif score >= 80:
    print("B학점")
else:
    print("C학점")
```

```python
# 3. 반복문 (for)
# C/Java: for(int i=0; i<5; i++)
# Python : for 변수 in range(횟수)
for i in range(5):  # 0부터 4까지 반복
    print(f"반복 {i}회차")
```

```python
# 4. 반복문 (while)
# 동작은 비슷하지만, 파이썬은 조건 뒤에 : 사용
count = 0
while count < 3:
    print(f"while 반복 {count}회차")
    count += 1
```

```python
# 5. 함수
# C/Java: 반환형과 매개변수 자료형 명시
# Python : def 함수명(매개변수):  → 자료형 명시 X
def add(a, b):
    """두 수를 더한 값을 반환"""
    return a + b

result = add(5, 7)
print("5 + 7 =", result)
```

```python
# 6. 리스트 & 리스트 내포
# C: 배열은 고정된 길이와 동일한 자료형
# Java: 배열은 고정 길이이고, List 인터페이스는 **제네릭(List<T>)**을 사용하여 다양한 타입의 요소 가능 but, 단일 자료형으로 제한
# Python: 크기 제한 없는 리스트 기본 제공, 서로 다른 자료형의 요소 가능

# 예시1: 파이썬 리스트는 여러 자료형을 함께 담을 수 있습니다.
mixed_list = ["apple", 100, 3.14, True]
print("예시1-", mixed_list,"\n")

# 예시2: 리스트 내포는 리스트를 더 간결하게 만듭니다.
fruits = ["apple", "banana", "cherry"]
# (일반적 방법)for 반복문을 사용해 각 요소를 대문자로 변환하고, 새로운 리스트에 추가.
fruits_upper = []
for fruit in fruits:
   fruits_upper.append(fruit.upper())
print("예시2-1(일반적)-", fruits_upper,"\n")

# (리스트 내포 사용)'fruits' 리스트의 각 'fruit'에 대해 'fruit.upper()'를 적용하여 새 리스트를 생성.
fruits_upper = [fruit.upper() for fruit in fruits]
print("예시2-2(리스트내포사용)-", fruits_upper,"\n")
```

```python
# 7. 딕셔너리 (Dictionary) - Python의 핵심 자료구조 중 하나
# C: 직접적인 내장 자료구조가 없어서 직접 구현해야 함.(보통 배열, 구조체, 해시테이블 라이브러리를 조합해서 흉내냄)
# Java: HashMap<String, Object> 와 유사
# Python: {key: value} 형태로 데이터를 저장 (key는 중복 불가, value는 어떤 자료형이든 가능)

person = {
    "name": "Alice",   # key: "name", value: "Alice"
    "age": 25,         # key: "age", value: 25
    "city": "Seoul"    # key: "city", value: "Seoul"
}
print("name:",person["name"],"\n")  # Java: map.get("name")

person["age"] = 26     # Java: map.put("age", 26)
print("age값 변경:",person,"\n")

person["email"] = "alice@example.com" # 새로운 key-value 추가
print("email 추가:",person,"\n")

if "city" in person:   # key 존재 여부 확인
    print("도시에 대한 정보가 있습니다.\n")

# key-value 삭제
del person["city"]     # key-value 삭제
print("삭제:",person,"\n")

# 딕셔너리 순회
for key, value in person.items(): #Java: for-each entrySet()
    print(key, ":", value)
```

```python
# 8. 예외 처리 (Exception Handling)
# - C: 예외 처리 구문이 없으며,보통 함수의 리턴값(0/음수)이나 errno를 확인하는 방식으로 에러 처리.
# - Java: try { ... } catch(Exception e) { ... }
# - Python: try: ... except Exception as e:

# 예시1-숫자로 변환 시도 → 실패 (문자 'a' 때문에 int 변환 불가)
try:
    num = int("100a")
    print("예시1-",end=" ")        # end 매개변수 사용- 끝에 공백(' ')을 추가하여 줄 바꿈 방지
    print("변환 성공:", num,"\n")  # end 매개변수 미사용 (기본값: '\n')

except ValueError as e:   # Java의 catch 블록과 비슷
    print("예시1-",end="")         # end 매개변수 사용- 끝에 아무것도 추가하지 않음 (빈 문자열)
    print("변환 실패:", e,"\n")

# 추가 예시: 0으로 나누기
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("예시2-",end="")         # end 매개변수 사용- 끝에 아무것도 추가하지 않음 (빈 문자열)
    print("0으로 나눌 수 없음:", e,"\n")

# 모든 예외 처리 (Java의 catch(Exception e) 와 동일)
try:
    x = undefined_variable
except Exception as e:
    print("예시3",end="-")       # end 매개변수 사용- 끝에 -을 추가
    print("예상치 못한 오류 발생:", e)
```

```python
# 9. 모듈 import
# C/Java: 외부 라이브러리 사용 시 import 또는 include
# Python : import 모듈명
import math
print("π 값:", math.pi)
print("제곱근(16):", math.sqrt(16))
```

```python
# NumPy와 Pandas 예제 코드
import numpy as np
import pandas as pd

# NumPy: 2x3 행렬 생성
a = np.array([[1, 2, 3], [4, 5, 6]])
print("NumPy 행렬:\n", a)
print("\n")

# Pandas: DataFrame 생성 및 평균 나이 계산
data = {
    '이름': ['철수', '영희'],
    '나이': [25, 28]
}
df = pd.DataFrame(data)
print("Pandas DataFrame:\n", df)
print(f"\n평균 나이: {df['나이'].mean()}")
```
