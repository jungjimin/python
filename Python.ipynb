🐍 파이썬 기본 문법 예제




```python
# 1. 변수와 자료형
# C/Java: int x = 10;  → 자료형(int) 명시 필수
# Python : 자료형 없이 변수명 = 값  → 동적 타이핑(dynamic typing)
# 정적 타이핑(Static Typing) → 변수의 **자료형(type)**을 코드 작성 시점에 미리 지정해야 함.
# 동적 타이핑(Dynamic Typing) → 변수의 자료형이 코드 실행 시점에 값에 따라 자동 결정됨.

x = 10              # 정수형(int)
y = 3.14            # 실수형(float)
name = "Alice"      # 문자열(str)
is_active = True    # 불리언(bool)
print(x, y, name, is_active)
```

```python
# 2. 조건문
# C/Java: 조건문에 () 필수, { }로 블록 표시
# Python : () 불필요, 들여쓰기로 블록 구분
score = 85
if score >= 90:
    print("A학점")
elif score >= 80:
    print("B학점")
else:
    print("C학점")
```

```python
# 3. 반복문 (for)
# C/Java: for(int i=0; i<5; i++)
# Python : for 변수 in range(횟수)
for i in range(5):  # 0부터 4까지 반복
    print(f"반복 {i}회차")
```

```python
# 4. 반복문 (while)
# 동작은 비슷하지만, 파이썬은 조건 뒤에 : 사용
count = 0
while count < 3:
    print(f"while 반복 {count}회차")
    count += 1
```

```python
# 5. 함수
# C/Java: 반환형과 매개변수 자료형 명시
# Python : def 함수명(매개변수):  → 자료형 명시 X
def add(a, b):
    """두 수를 더한 값을 반환"""
    return a + b

result = add(5, 7)
print("5 + 7 =", result)
```

```python
# 6. 리스트 & 리스트 내포
# C/Java: 배열은 고정 길이 / List는 제네릭 사용
# Python : 크기 제한 없는 리스트 기본 제공
fruits = ["apple", "banana", "cherry"]
fruits_upper = [fruit.upper() for fruit in fruits]  # 리스트 내포
print(fruits_upper)
```

```python
# 7. 딕셔너리
# C/Java: Map<String, Object> 와 비슷
# Python : {key: value} 구조
person = {"name": "Alice", "age": 25, "city": "Seoul"}
print(person["name"])
person["age"] = 26
print(person)
```

```python
# 8. 예외 처리
# C/Java: try { ... } catch(Exception e) { ... }
# Python : try: ... except Exception as e:
try:
    num = int("100a")  # 숫자로 변환 시도 (실패)
except ValueError as e:
    print("변환 실패:", e)
```

```python
# 9. 모듈 import
# C/Java: 외부 라이브러리 사용 시 import 또는 include
# Python : import 모듈명
import math
print("π 값:", math.pi)
print("제곱근(16):", math.sqrt(16))
```

```python
# 10. 파일 입출력
# C/Java: 파일 객체 열기 → 읽기/쓰기 → 닫기 필요
# Python : with문 사용 시 자동 close()
with open("sample.txt", "w", encoding="utf-8") as f:
    f.write("안녕하세요, 파이썬!\n")

with open("sample.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print("파일 내용:", content)
```
