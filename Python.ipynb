# 2. 파이썬 특징

파이썬의 장점은 크게 여섯 가지로 정리할 수 있습니다.

1. 생산성 – “인생은 짧으니 파이썬이 필요하다.”라는 말처럼 개발 속도가 빠릅니다.

2. 풍부한 라이브러리 – 내장 라이브러리가 많고 확장성도 뛰어납니다.

3. 가독성 – 코드가 단순하고 읽기 쉽습니다.

4. 접착성 – C 언어 모듈 등을 쉽게 붙여 쓸 수 있습니다.

5. 무료 – 파이썬 소프트웨어 재단에서 관리하며 무료로 제공됩니다.

6. 유니코드 지원 – 한글, 영어, 중국어 등 다양한 언어를 문제없이 처리할 수 있습니다.

또한, 파이썬은 동적 타이핑 언어라서 실행 시 타입을 자동으로 체크하고 메모리를 관리해 줍니다. 현재 미국 상위 대학의 컴퓨터 학과 대부분이 첫 프로그래밍 언어로 파이썬을 채택하고 있습니다

# 7. 파이썬 기본 문법

```python
# 1. 변수와 자료형
# C/Java: int x = 10;  → 자료형(int) 명시 필수
# Python : 자료형 없이 변수명 = 값  → 동적 타이핑(dynamic typing)
# 정적 타이핑(Static Typing) → 변수의 **자료형(type)**을 코드 작성 시점에 미리 지정해야 함.
# 동적 타이핑(Dynamic Typing) → 변수의 자료형이 코드 실행 시점에 값에 따라 자동 결정됨.

x = 10              # 정수형(int)
y = 3.14            # 실수형(float)
name = "Alice"      # 문자열(str)
is_active = True    # 불리언(bool)
print(x, y, name, is_active)
```

```python
# 2. 조건문
# C/Java: 조건문에 () 필수, { }로 블록 표시
# Python : () 불필요, 들여쓰기로 블록 구분
score = 85
if score >= 90:
    print("A학점")
elif score >= 80:
    print("B학점")
else:
    print("C학점")
```

```python
# 3. 반복문 (for)
# C/Java: for(int i=0; i<5; i++)
# Python : for 변수 in range(횟수)
for i in range(5):  # 0부터 4까지 반복
    print(f"반복 {i}회차")
```

```python
# 4. 반복문 (while)
# 동작은 비슷하지만, 파이썬은 조건 뒤에 : 사용
count = 0
while count < 3:
    print(f"while 반복 {count}회차")
    count += 1
```

```python
# 5. 함수
# C/Java: 반환형과 매개변수 자료형 명시
# Python : def 함수명(매개변수):  → 자료형 명시 X
def add(a, b):
    """두 수를 더한 값을 반환"""
    return a + b

result = add(5, 7)
print("5 + 7 =", result)
```

```python
# 6. 리스트 & 리스트 내포
# C/Java: 배열은 고정 길이 / List는 제네릭 사용
# Python : 크기 제한 없는 리스트 기본 제공
fruits = ["apple", "banana", "cherry"]
fruits_upper = [fruit.upper() for fruit in fruits]  # 리스트 내포
print(fruits_upper)
```

```python
# 7. 딕셔너리
# C/Java: Map<String, Object> 와 비슷
# Python : {key: value} 구조
person = {"name": "Alice", "age": 25, "city": "Seoul"}
print(person["name"])
person["age"] = 26
print(person)
```

```python
# 8. 예외 처리
# C/Java: try { ... } catch(Exception e) { ... }
# Python : try: ... except Exception as e:
try:
    num = int("100a")  # 숫자로 변환 시도 (실패)
except ValueError as e:
    print("변환 실패:", e)
```

```python
# 9. 모듈 import
# C/Java: 외부 라이브러리 사용 시 import 또는 include
# Python : import 모듈명
import math
print("π 값:", math.pi)
print("제곱근(16):", math.sqrt(16))
```

```python
# 10. 파일 입출력
# C/Java: 파일 객체 열기 → 읽기/쓰기 → 닫기 필요
# Python : with문 사용 시 자동 close()
with open("sample.txt", "w", encoding="utf-8") as f:
    f.write("안녕하세요, 파이썬!\n")

with open("sample.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print("파일 내용:", content)
```

# 8. 파이썬 핵심 라이브러리

**(1) 데이터 분석 라이브러리: NumPy & Pandas**

파이썬이 데이터 분석 분야에서 필수적인 이유! 바로 NumPy와 Pandas 덕분입니다.

**NumPy: 수치 연산**에 특화된 라이브러리입니다. 다차원 배열(행렬) 연산을 빠르고 효율적으로 처리합니다.

**Pandas: 정형 데이터 분석**에 최적화된 라이브러리입니다. 엑셀처럼 표 형태의 데이터를 쉽게 다루고 분석할 수 있습니다.

📌 주요 활용 사례

**구글/넷플릭스**: 머신러닝 모델의 복잡한 수치 연산에 NumPy를 활용합니다.

**우버/JP모건**: 방대한 고객 데이터나 금융 데이터를 Pandas로 분석해 비즈니스 결정을 내립니다.

```python
# NumPy와 Pandas 예제 코드
import numpy as np
import pandas as pd

# NumPy: 2x3 행렬 생성
a = np.array([[1, 2, 3], [4, 5, 6]])
print("NumPy 행렬:\n", a)
print("\n")

# Pandas: DataFrame 생성 및 평균 나이 계산
data = {
    '이름': ['철수', '영희'],
    '나이': [25, 28]
}
df = pd.DataFrame(data)
print("Pandas DataFrame:\n", df)
print(f"\n평균 나이: {df['나이'].mean()}")
```
